```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()

lemmatizer.lemmatize("going",'v')
```

```python
'''
POS - Noun-n
verb - v
adjective-a
adverb - r
'''
words = ["eating","eats" ,"eaten","writing","writes","programming","history"]
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,"v"))
```
